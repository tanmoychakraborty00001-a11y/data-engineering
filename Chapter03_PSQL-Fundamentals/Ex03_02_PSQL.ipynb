```python
def print_header(title):
    print(f"\n{'='*60}")
    print(f"📋 {title}")
    print('='*60)
```

```python
# 1. CREATE TABLE 

def create_table():
    print_header("1. CREATE TABLE")
    try:
        conn = connect_db()
        cur = conn.cursor()
        cur.execute("""
        DROP TABLE IF EXISTS salesnew1;
        CREATE TABLE salesnew1 (
        order_id        VARCHAR(50) PRIMARY KEY,
        country         VARCHAR(50),
        order_value_eur NUMERIC(12,2),
        cost            NUMERIC(12,2),
        order_date      DATE,
        category        VARCHAR(50),
        customer_name   VARCHAR(100),
        sales_manager   VARCHAR(50),
        sales_rep       VARCHAR(50),
        device_type     VARCHAR(30)
            );
        
        """)
        conn.commit()
        print("✅ CREATE TABLE salesnew1 - SUCCESS")
        cur.close()
        conn.close()
    except Exception as e:
        print(f"❌ Error: {e}")
```

```python
def show_all_tables():
    """Show ALL data from ALL tables"""
    # 1. Keeping Fancy header name
    print_header("ALL TABLES - COMPLETE DATA")
    
    # 2. Open DB connection
    conn = connect_db()

    # 3. List down all tables
    cur = conn.cursor()
    cur.execute("SELECT table_name FROM information_schema.tables WHERE table_schema='public';")
    
    # 4. Extract table names
    tables = [row[0] for row in cur.fetchall()]
    print(f"📋 Tables found: {tables}")

    # 5. Show all data from each table
    for table in tables:
        # 6. Creating Table header
        print(f"\n📊 {table} (ALL {table.upper()})")
        print("-" * 50)
        # 7. SELECT ALL from table
        cur.execute(f"SELECT * FROM {table};")
        # 8. Get All rows
        rows = cur.fetchall()
        for row in rows:
            # 9. Print each row 
            print(f" {row}")
    # 10. Clean up cursor
    cur.close()
    # 11. Closing connection
    conn.close()
```

```python
show_all_tables()
```
