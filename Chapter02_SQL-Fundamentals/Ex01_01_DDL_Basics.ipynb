```python
conn.execute("""SELECT SUM(CAST(REPLACE(order_value_EUR, ',', '') AS DOUBLE)) AS total_sales_eur FROM SALES""").df()
```

```python
conn.execute("""SELECT category, SUM(CAST(REPLACE(order_value_EUR, ',', '') AS DOUBLE)) AS total_sales_eur FROM SALES GROUP BY category 
ORDER BY total_sales_eur DESC; """).df()
```

```python
conn.execute("""SELECT category, SUM(CAST(REPLACE(order_value_EUR, ',', '') AS DOUBLE)) AS Revenue, SUM(cost) AS Total_cost, 
SUM(CAST(REPLACE(order_value_EUR, ',', '') AS DOUBLE)) - SUM(cost) AS Profit FROM sales GROUP BY category ORDER BY Profit DESC;
""").df()
```
